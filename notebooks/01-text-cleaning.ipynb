# Text Cleaning Task — text_sample.txt

**Goal:** Clean a deliberately messy text file (accents, emojis, HTML, inconsistent prices, contractions, symbols, extra spaces) while preserving its meaning, and save the result.

## Step 1: Read & inspect
Read the whole file into one string with `open().read()` and print it to see the mess before cleaning.


```python
with open('text_sample.txt', 'r', encoding='utf-8') as file:
    text = file.read()

print(text)
```

## Step 3: Fix encoding, remove HTML, expand contractions
Order matters: fix encoding/accents first (clean characters for every later step), then strip HTML tags, then expand contractions — done before punctuation removal because it needs the apostrophe in "can't".


```python
import re
import string
import ftfy
import unicodedata
import contractions

# 1. Fix encoding & strip accents FIRST
text = ftfy.fix_text(text)
text = unicodedata.normalize('NFKD', text).encode('ascii', 'ignore').decode()

# 2. Remove HTML tags
text = re.sub(r'<.*?>', '', text)

# 3. Expand contractions (BEFORE punctuation removal - needs the apostrophe)
text = contractions.fix(text)

print(text)
```

## Step 4: Standardize prices

The file writes the same price three ways ($29.99, 29.99 USD, 29,99$). Unify them to one canonical form before lowercasing/punctuation removal, which would otherwise break the pattern.


```python
# 4. Standardize the three price formats into one
text = re.sub(r'\$?29[.,]99\s*(USD|\$)?', '$29.99', text)

print(text)
```

## Step 5: Final passes — lowercase, punctuation, numbers, whitespace
Lowercase for consistency, strip all punctuation/symbols (@@, ##, repeated !!), remove numbers, then collapse whitespace LAST to clean up the gaps left by every removal above.


```python
# 5. Lowercase
text = text.lower()

# 6. Remove punctuation and random symbols (!!, ,,,,, @@, ##)
text = text.translate(str.maketrans('', '', string.punctuation))

# 7. Remove numbers
text = re.sub(r'\d+', '', text)

# 8. Collapse extra spaces & newlines LAST
text = re.sub(r'\s+', ' ', text).strip()

print(text)
```

## Step 6: Save the cleaned text
Write the cleaned string to a new file `cleaned_text.txt` in write mode.


```python
# Save the cleaned text to a new file
with open('cleaned_text.txt', 'w', encoding='utf-8') as file:
    file.write(text)

print("Saved cleaned_text.txt")
```
