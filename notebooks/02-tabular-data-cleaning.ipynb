# Loan Data Cleaning — bank_loan.csv

**Dataset:** bank_loan.csv (5,000 rows × 14 columns), bank personal loan data from Kaggle.

**Goal:** Inspect and clean the dataset so it's accurate, consistent, and ready for analysis.

**Setup:** Import pandas, then upload the CSV and read it into a DataFrame (`df`).



```python
# Importing pandas
import pandas as pd
```

```python
# Upload the CSV file and reading it into a DataFrame
from google.colab import files
uploaded = files.upload()

df = pd.read_csv('bank_loan.csv')
```

## Step 1: Inspect the data
Use `head()` to eyeball values and `info()` to check structure, types, and non-null counts before deciding what to clean.

**Observations:**
- 5,000 rows × 14 columns; all columns 5,000 non-null → no missing values.
- 13 columns are numeric (int64); **CCAvg is `object` (text)** — the odd one out.
- CCAvg values look like `1/60`, `1/50` — slashes where decimals should be. Flag to fix.


```python
# Inspect the first few rows of DataFrame
print("First five rows of the dataset:")
print(df.head())

# Get a summary of the DataFrame
print("\nDataFrame summary: ")
print(df.info())
```

## Step 2: Check missing values & duplicates
- `isnull().sum()` → **0 missing** in all 14 columns.
- `duplicated().sum()` → **0 duplicate rows**.

No action needed for either — the dataset is complete and has no repeats. The only real issue is CCAvg's format (fixed next).


```python
# Detecting and handling missing values

#print(df.isnull())
print("NUll value Report:\n \n", df.isnull().sum())

duplicates = df.duplicated()
print("\nDuplicate Value found:", duplicates.sum())
```

## Step 3: Fix CCAvg data type
CCAvg held values like `1/60` and was stored as text (`object`), so no math or modeling could use it. The slash is a mis-encoded decimal point.

**Action:** Replaced `/` with `.`, then converted to `float`.

**Result:** CCAvg is now `float64` (`1.6`, `1.5`, `1.0`…) — a usable number. Chose float over int because values are decimals.


```python
# Fix CCAvg: slashes should be decimal points, then convert text -> number
df['CCAvg'] = df['CCAvg'].str.replace('/', '.', regex=False)  # "1/60" -> "1.60"
df['CCAvg'] = df['CCAvg'].astype(float)                        # "1.60" -> 1.60

# Verify
print(df['CCAvg'].dtype)
print(df['CCAvg'].head())
```

## Step 4: Final check & save
Re-ran `info()` to confirm the fix: CCAvg is now `float64`, all 14 columns 5,000 non-null. Dataset is clean.

Saved the result as `bank_loan_cleaned.csv` with `index=False` (so pandas' row numbers aren't written as an extra column).


```python
df.info()   # confirm CCAvg is float64 and all looks right
```

```python
df.to_csv('bank_loan_cleaned.csv', index=False)
```
