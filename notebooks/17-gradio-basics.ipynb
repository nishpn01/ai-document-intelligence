```python
import gradio as gr
import math

def math_powers(number):
    n = int(number)                      # gr.Number gives a float; factorial needs an int
    square = n ** 2
    cube = n ** 3
    factorial = math.factorial(n)        # built-in — no manual loop
    return square, cube, factorial

demo = gr.Interface(
    fn=math_powers,
    inputs=gr.Number(label="Enter a number"),
    outputs=[
        gr.Number(label="Square"),
        gr.Number(label="Cube"),
        gr.Number(label="Factorial"),
    ],
    title="Math Superpowers",
    description="Enter a whole number to get its square, cube, and factorial."
)

demo.launch(share=True)
```
