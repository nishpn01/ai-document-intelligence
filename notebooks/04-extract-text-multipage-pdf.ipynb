```python
import fitz
import json

doc = fitz.open("your-document.pdf")
results = []

for page_num in range(len(doc)):
    page = doc[page_num]
    for w in page.get_text("words"):
        x0, y0, x1, y1, word, block_no, line_no, word_no = w
        results.append({
            "page": page_num + 1,
            "text": word,
            "bbox": [x0, y0, x1, y1],
            "block": block_no,
            "line": line_no,
            "word": word_no
        })

print(f"Extracted {len(results)} words across {len(doc)} pages")
print(json.dumps(results[:5], indent=2))   # preview first 5
```

```python
with open("sdf_extracted.json", "w") as f:
    json.dump(results, f, indent=2)
files.download("sdf_extracted.json")
```
